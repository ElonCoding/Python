Q.1 Write a program to print Fizz if the no is divisible by 3 Buzz if the no is divisible by 5 then NO FizzBuzz if the no is divisible by 3 and 5 then print no one upto 30

````python
# Write a program to print Fizz if the no is divisible by 3 Buzz if the no is divisible by 5 then NO FizzBuzz if the no is divisible by 3 and 5 then print no one 
for i in range(1, 30):
     if i % 3 == 0 and i % 5 == 0:
         print("FizzBuzz")
     elif i % 3 == 0:
         print("Fizz")
     elif i % 5 == 0:
         print("Buzz")
     else:
         print(i)
````

````python
# Another way to do the same thing but by taking input from user

num = int(input("Enter a number: "))
if num % 3 == 0 and num % 5 == 0:
    print("FizzBuzz")
elif num % 3 == 0:
    print("Fizz")
elif num % 5 == 0:
    print("Buzz")
else:
    print(num)
````

Q.2 Write a program which will take number form user and print all the factors or that 

````python
#Q.2 Write a program which will take number form user and print all the factors or that number
num = int(input("Enter a number: "))
print("Factors of", num, "are:")
for i in range(1, num + 1):
    if num % i == 0:
        print(i)
````

Q.3 which will take a no as the print count of all the even and odd integers in the count 

````python
#which will take a no as the print count of all the even and odd integers in the count 
num = int(input("Enter a number: " ))
even_count = 0
odd_count = 0
for i in range(1, num + 1):
    if i % 2 == 0:
        even_count += 1
    else:
        odd_count += 1
print("Count of even numbers:", even_count)
print("Count of odd numbers:", odd_count)
````

Q.4 Write a program to check whether the number is prime or not using for loop 

````python
# Write a program to check whether the number is prime or not using for loop 


num = int(input("Enter a number: "))
if num > 1:
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            print(num, "is not a prime number")
            break
    else:
        print(num, "is a prime number")
````

Q.5 write a progeam which will take a num user and count no. of digit persent in that num

````python
#write a program which will take a number from the user and count the number of digits present in that number
num = int(input("Enter a number: "))
count = 0
while num > 0:
    num //= 10
    count += 1
print("Number of Digits:", count)
````

Q.6 Write a program to Sum of all digits for user
<br>

Input: 123
<br>
Output: 6
```

```

````python
#Write a program to Sum of all digits for user
num = int(input("Enter a number: "))
ds = 0   
while num > 0:
    ds += num % 10
    num //= 10
print("Sum of digits:", ds)
````

Q.7 write a program to reverse the number provided by user
<br>
Example <br>
Input: 123
<br>
Output: 321

````python
#write a program to reverse the number provided by user
num = int(input("Enter a number: "))
rn = 0
while num > 0:
    digit = num % 10
    rn = rn * 10 + digit
    num //= 10
print("Reversed number:", rn)
````
